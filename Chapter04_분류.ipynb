### 앙상블 학습 - 보팅 분류기 (Voting Classifier)

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

from sklearn.ensemble import VotingClassifier # 앙상블 학습 기법 중 하나
from sklearn.linear_model import LogisticRegression # 로지스틱 회귀 모델
from sklearn.neighbors import KNeighborsClassifier # K-최근접 이웃 분류 모델
from sklearn.datasets import load_breast_cancer # 유방암 데이터셋
from sklearn.model_selection import train_test_split # 데이터셋을 훈련 세트와 테스트ㅡ 세트로 분할
from sklearn.metrics import accuracy_score #
cancer = load_breast_cancer()
```

```python
# /content/drive/MyDrive/kwu/ML/data/Breast_Cancer.csv
```

```python
data_df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
print('피처 수:', data_df.shape[1])
data_df.head(3)
```

```python
# 소프트 보팅 기반 앙상블 구성
lr_clf = LogisticRegression(solver='liblinear', random_state=156) # 로지스틱 회귀 분류기 객체
knn_clf = KNeighborsClassifier(n_neighbors=8) # K-최근접 이웃 분류기 객체

vo_clf = VotingClassifier(#VotingClassifier 객체
                          estimators=[('LR', lr_clf), ('KNN', knn_clf)],
                          voting ='soft' # 보팅 방식
                          )

# 유방암 데이터로 훈련 세트와 테스트 세트로 분할
X_train, X_test, y_train, y_test = train_test_split(cancer.data, cancer.target, test_size=0.2, random_state=156)

# 보팅 앙상블 학습/평가
vo_clf.fit(X_train, y_train) # 보팅 분류기(앙상블 모델)을 학습
pred = vo_clf.predict(X_test) # 학습된 보팅 분류기를 사용하여 예측을 수행
print('Voting 분류기 정확도: {0:.4f}'.format(accuracy_score(y_test, pred)))

# 개별 모델 성능과 비교
print('\n[개별 모델 성능 비교]')
for clf in [lr_clf, knn_clf]:
    clf.fit(X_train, y_train)
    pred = clf.predict(X_test)
    print(f'{clf.__class__.__name__}: {accuracy_score(y_test, pred):.4f}')
```

### 랜덤 포레스트 (Random Forest)

```python
# 사용자 행동 인식 데이터로 실습
# https://archive.ics.uci.edu/contribute/donation
```

```python
# UCI 사용자 행동 인식 데이터셋을 로드

feature_name_df = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/features.txt', sep='\s+',
                              header=None, names=['column_index','column_name'])
feature_name_df
```

```python
feature_dup_df = pd.DataFrame(
      data=feature_name_df.groupby('column_name').cumcount(), #각 그룹 내에서 중복 횟수를 카운트
      columns=['dup_cnt'] # 중복 횟수 컬럼의 이름
)
feature_dup_df = feature_dup_df.reset_index()
feature_dup_df
```

```python
new_feature_name_df = pd.merge(feature_name_df.reset_index(),feature_dup_df, how='outer')
# 중복 횟수가 0보다 크면 '원래이름_중복횟수', 아니면 '원래이름'
new_feature_name_df['column_name'] = new_feature_name_df[['column_name','dup_cnt']].apply(
    lambda x: x[0]+'_'+str(x[1]) if x[1] > 0 else x[0], axis=1
)
new_feature_name_df = new_feature_name_df.drop(['index'], axis=1) # 컬럼 삭제
new_feature_name_df
```

```python
import numpy as np
# Get the 'column_name' series which should have unique names after previous processing
feature_name_series = new_feature_name_df['column_name']

# Convert to list, filtering out any NaN values and ensuring they are strings
feature_name = [str(x) for x in feature_name_series.dropna().tolist() if pd.notna(x)]

# Ensure absolute uniqueness by converting to a set and back to list (preserves order)
feature_name = list(dict.fromkeys(feature_name))

feature_name
```

```python
# 학습/테스트 피처 데이터 (sep=공백, 컬럼명 적용)
X_train = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/train/X_train.txt', sep='\s+', names=feature_name, header=None)
X_test = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/test/X_test.txt', sep='\s+', names=feature_name, header=None)

# 학습/테스트 레이블 (행동 종류: 1~6)
y_train = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/train/y_train.txt', sep='\s+', header=None, names=['action'])
y_test = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/test/y_test.txt', sep='\s+', header=None, names=['action'])
```

```python
print('학습셋:', X_train.shape, '테스트셋:', X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 기본 랜덤 포레스트 학습
rf_clf = RandomForestClassifier(random_state=156)
rf_clf.fit(X_train, y_train)
pred = rf_clf.predict(X_test)
print('랜덤 포레스트 기본 정확도: {0:.4f}'.format(accuracy_score(y_test, pred)))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# 피처 중요도 시각화

ftr_importance = pd.Series(
    rf_clf.feature_importances_,
    index=X_train.columns
)
ftr_top20 = ftr_importance.sort_values(ascending=False)[:20] # 피처 중요도 내림차순 정렬, 상위 20개의 피처 선택

plt.figure(figsize=(9,6))
plt.title('Random Forest Feature Importance Top 20')
sns.barplot(x=ftr_top20, y=ftr_top20.index, palette='viridis')
plt.tight_layout()
plt.show()
```

```python
ftr_importance
```

```python
from sklearn.model_selection import GridSearchCV
# GridSearchCV로 최적 하이퍼파라미티 탐색

params = {
    'n_estimators': [100],
    'max_depth': [6, 8, 10, 12],
    'min_samples_leaf': [8, 12, 18],
    'min_samples_split': [8, 16, 20],
}
rf_clf_gs = RandomForestClassifier(random_state=0, n_jobs=-1) # n_jobs=-1 : 모든 CPU 코어 사용 (병렬 처리)
grid_cv = GridSearchCV(rf_clf_gs, param_grid=params, cv=2, n_jobs=-1)
y_train_reshaped = y_train.values.ravel() # 1차원 배열로 평탄화, GridSearchCV가 y 데이터를 처리하는 방식에 필요
grid_cv.fit(X_train, y_train_reshaped)

print('최적 하이퍼 파라미터:\n',grid_cv.best_params_) # 최적의 하이퍼파라미터 조합
print(f'최고 예측 정확도: {grid_cv.best_score_}')
```

```python
# 최적 파라미터로 최종 학습
rf_clf1 = RandomForestClassifier(
    n_estimators = 300, # 트리 개수를 300개
    max_depth = 10, # 'mx_depth'를 'max_depth'로 수정
    min_samples_leaf=8,
    random_state=0
)
y_train_reshaped = y_train.values.ravel()
rf_clf1.fit(X_train, y_train_reshaped)
pred = rf_clf1.predict(X_test)
print(f'최적 파라미터 랜덤 포레스트 정확도: {accuracy_score(y_test,pred):.4f}')
```

### GBM (Gradient Boosting Machine)

```python
from sklearn.ensemble import GradientBoostingClassifier
import time

# 사용자 행동 인식 데이터셋 사용
start_time = time.time()
gb_clf = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate = 0.1,
    max_depth=3,
    random_state=0
)
gb_clf.fit(X_train, y_train_reshaped)
gb_pred= gb_clf.predict(X_test)

print(f'GBM 정확도: {accuracy_score(y_test, gb_pred)}')
print(f'GBM 수행 시간: {time.time() - start_time}') # GBM은 순차 학습 -> 랜덤 포레스트보다 훨씬 느림
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import GridSearchCV
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score # Needed for the final accuracy print statement

# --- Start of code to define X_train, y_train_reshaped, X_test for Human Activity Recognition dataset ---
# This section is added to ensure variables are defined if preceding cells were not executed.

# Load feature names (from pKY76afQaAUx)
feature_name_df = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/features.txt', sep='\s+',
                              header=None, names=['column_index','column_name'])

# Handle duplicate feature names (from LecexVWpaKmU, iSVvDevxaOAz)
feature_dup_df = pd.DataFrame(
      data=feature_name_df.groupby('column_name').cumcount(), #각 그룹 내에서 중복 횟수를 카운트
      columns=['dup_cnt'] # 중복 횟수 컬럼의 이름
)
feature_dup_df = feature_dup_df.reset_index()

new_feature_name_df = pd.merge(feature_name_df.reset_index(),feature_dup_df, how='outer')
new_feature_name_df['column_name'] = new_feature_name_df[['column_name','dup_cnt']].apply(
    lambda x: x[0]+'_'+str(x[1]) if x[1] > 0 else x[0], axis=1
)
new_feature_name_df = new_feature_name_df.drop(['index'], axis=1)

# Create final feature_name list (from 7cTafwnhaT08)
feature_name_series = new_feature_name_df['column_name']
feature_name = [str(x) for x in feature_name_series.dropna().tolist() if pd.notna(x)]
feature_name = list(dict.fromkeys(feature_name))

# Load training/test data and labels (from p9wcD0Hka3Ks)
X_train = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/train/X_train.txt', sep='\s+', names=feature_name, header=None)
X_test = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/test/X_test.txt', sep='\s+', header=None, names=feature_name)
y_train = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/train/y_train.txt', sep='\s+', header=None, names=['action'])
y_test = pd.read_csv('/content/drive/MyDrive/kwu/ML/data/human_activity/UCI HAR Dataset/test/y_test.txt', sep='\s+', header=None, names=['action'])

# Reshape y_train (from zOPeAokZe6EJ)
y_train_reshaped = y_train.values.ravel()
# --- End of added data loading/preprocessing code ---

# GBM GridSearchCV (시간이 오래 걸림 - 참고용)
params = {
    'n_estimators': [100, 500],
    'learning_rate': [0.05, 0.1]
}
gb_clf_gs = GradientBoostingClassifier(random_state=0)
grid_cv = GridSearchCV(gb_clf_gs, param_grid=params, cv=2, verbose=1)
grid_cv.fit(X_train, y_train_reshaped)

print('최적 파라미터:', grid_cv.best_params_)
print(f'최고 정확도: {grid_cv.best_score_}')

gv_pred = grid_cv.best_estimator_.predict(X_test) # .best_estimator 속성에서 최적 모델을 가져오ㅏ서
print(f'GBM GridSearchCV 예측 정확도: {accuracy_score(y_test, gv_pred):.4f}')
```
